# Estudo sobre Modelos Generativos de Deep Learning: GANs e VAEs

## 1. Introdução

Modelos generativos em deep learning aprendem a distribuição dos dados de entrada, possibilitando gerar exemplos plausíveis e inéditos. Os principais paradigmas atuais são as Generative Adversarial Networks (GANs) e os Variational Autoencoders (VAEs).

## 2. Generative Adversarial Networks (GANs)

**Conceito:** GANs foram propostas por Goodfellow et al. (2014). Consistem em dois componentes: um Gerador (Generator), que cria dados sintéticos a partir de ruído, e um Discriminador (Discriminator), que tenta diferenciar entre dados reais e gerados. O treinamento é um jogo minimax.

**Arquiteturas avançadas:**
- DCGAN: GANs profundas com convoluções.
- StyleGAN2: Geração de faces realistas e controle de estilo.
- CycleGAN: Tradução de imagens entre domínios sem pares.

**Desafios:**
- Instabilidade no treinamento
- Mode collapse (geração limitada de diversidade)
- Métricas de avaliação (Inception Score, FID)

**Aplicações:**
- Geração de imagens realistas
- Super-resolução
- Transferência de estilo
- Data augmentation

**Exemplos e modelos:**
- [DCGAN - PyTorch](https://github.com/pytorch/examples/tree/main/dcgan)
- [StyleGAN2-ADA - Nvidia](https://github.com/NVlabs/stylegan2-ada-pytorch)
- [CycleGAN - PyTorch](https://github.com/junyanz/pytorch-CycleGAN-and-pix2pix)

## 3. Variational Autoencoders (VAEs)

**Conceito:** VAEs (Kingma & Welling, 2013) são autoencoders probabilísticos, onde o encoder gera parâmetros de uma distribuição latente (tipicamente gaussiana) e o decoder reconstrói a entrada a partir de amostras dessa distribuição. O objetivo é maximizar a evidência inferior (ELBO).

**Extensões:**
- β-VAE: maior disentanglement no espaço latente
- VQ-VAE: espaço latente discreto, base de modelos como DALL-E

**Desafios:**
- Amostras menos nítidas que GANs
- Equilíbrio entre regularização e reconstrução

**Aplicações:**
- Geração e interpolação de imagens/áudio/texto
- Compressão e redução de dimensionalidade
- Análise de anomalias

**Exemplos e modelos:**
- [VAE - PyTorch](https://github.com/pytorch/examples/tree/main/vae)
- [Beta-VAE](https://github.com/1Konny/Beta-VAE)
- [VQ-VAE-2 - PyTorch](https://github.com/rosinality/vq-vae-2-pytorch)

## 4. Comparação rápida

| Aspecto               | GANs                              | VAEs                                 |
|-----------------------|------------------------------------|--------------------------------------|
| Realismo das amostras | Alto                               | Moderado                             |
| Espaço latente        | Pouco estruturado                  | Estruturado e navegável              |
| Treinamento           | Instável, sensível a hiperparâmetros| Estável, probabilístico              |
| Aplicações            | Geração realista, transferência    | Interpolação, compressão, anomalias  |

## 5. Perspectivas recentes
- **Modelos de difusão:** combinam vantagens dos dois paradigmas e geram imagens de alta qualidade (ex: Stable Diffusion, DALL-E 2)
- **Multimodalidade:** geração de texto+imagem
- **Ética:** deepfakes, desinformação, responsabilidade
